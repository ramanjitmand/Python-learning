```python
# A function allows you to reuse a piece of code parsing whatever parameters to it so you can do it multiple times or do it multiple times with different parameters
# Use function if you know you need to do it again and again
# Let's write a function - fn is function name
# Indentations are part of function (4 spaces or 1 tab is indentation)
def fn():
    print('Hello, world!')

# You have to do next line for it to work - 'call' the function
fn()
```
