```python
import pandas as pd

# variable is url
url = r"https://raw.githubusercontent.com/data-to-insight/ERN-sessions/main/data/1980%202023%20average%20house%20prices.csv"

# Use read_csv function from pandas
df = pd.read_csv(url)

df.head()
```

```python
df.info()
# This helps us get a gist of the data
```

```python
# Let's make all columns lowercase as we see uppercase halfway through if we just do df.info() - easier to work with that way
# df.columns gives us column headings
# df.columns = str.lower
# It'll take every bit of list series and make it lowercase then overwrite the original column names
df.columns = map(str.lower, df.columns)

df.info()
# This helps us get a gist of the data
```

```python
# Make 'period' column into datetime

# coerce gives errors as NA rather than ignorning them
df["period_dt"] = pd.to_datetime(df['period'], format="%Y-%m", errors='coerce')

df.info()
```

```python
# plotly express - allows you to make a plot quickly and easily; this is a graph objects - it's a submodule of graph objects module
# graph objects - really powerful but slightly long winded making it e.g. moving titles around, orders, colours etc

# px is short for plot(ly) ex(press)
import plotly.express as px

# Scatterplot is good to show time series data
price_over_time = px.scatter(df,
                             x="period_dt",
                             y="average price all property types")

price_over_time.show()

# You can zoom in/out of certain bits (hover on RHS of plot)
# If you double click you can go back to the centre
# Lasso select - allows you to make a selection of data in plot itself and investigate it further and change a different plot - advanced stuff!
```

```python
# Let's make a title and x-axis and y-axis names, plus fit a trend line to see if there's a trend/relationship between x and y
# Hit control + j to open your terminal, or click 3 lines on top LHS and get terminal from there
# In terimanl type pip install statsmodels then click x in popup to close it down

import statsmodels

# We have to use period_dt because it won't work if we use period (which isn't in the format - time - that scatter plot needs it to be)
# New visual where ols will look at a trend line
# Labels bit - LHS gives us original name and RHS of : gives us what we want it to say instead
# ols is a type of trend line? There are others out there?
# ols - order of least squares gives a line closest to the points
price_over_time_ols = px.scatter(df, 
                                 x='period_dt',
                                 y='average price all property types',
                                 title="Change in average property price (Monthly/£)",
                                 labels={'period_dt':"Time period",
                                         'average price all property types':"House Prices"},
                                 trendline="ols")

price_over_time_ols.show()
```
