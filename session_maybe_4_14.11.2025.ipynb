```python
# Let's use pandas (installing is harder than importing)
# Let's name it pd because we're going to keep using pandas (it's universal to name pandas as pd - keep alises same when wriitng code for other things you import so someone can read your code)
import pandas as pd

# You can read files from your operating system (os)
import os

# We'll be taking from GitHub 
# r means Python read every character as a string exactly as it is

ci_filepath = r"https://raw.githubusercontent.com/data-to-insight/ERN-sessions/main/data/ChildIdentifiers.csv"

cc_filepath = r"https://raw.githubusercontent.com/data-to-insight/ERN-sessions/main/data/ChildCharacteristics.csv"

# From pandas let's read the csv (need . to say to do something)
identifiers = pd.read_csv(ci_filepath)
# You can do header, skiprows etc

characteristics = pd.read_csv(cc_filepath)
```

```python
# .info is main way to see info
identifiers.info()

# If Dtype comes up 'object' this is basically a string (DOB ought to be a date not string)
characteristics.info()

# .head method shows top few rows 
identifiers.head()
# Standard is 5 but you can go up to 20 if you parse

# Let's make DOB a date type
# Let's not overwrite DOB column we already have - make a new column
# Square brackets mean make a selection
# Let's make a new PersonBirthDate_dt column with as a date type column
# Single brackets makes that thing a string
# strftime - you can Google this (covered last week); capital means 4 digits, lowercase is 2 digits
# Snake case inc _ (used for method/function/variable)
identifiers['PersonBirthDate_dt'] = pd.to_datetime(identifiers['PersonBirthDate'], format="%Y-%m-%d")

identifiers.info()

identifiers['ExpectedPersonBirthDate_dt'] = pd.to_datetime(identifiers['PersonBirthDate'], format="%Y-%m-%d")

# Adding errors means you can tackle if date doesn't fit this style in original dataset
identifiers['PersonDeathDate_dt'] = pd.to_datetime(identifiers['PersonDeathDate'], format="%Y-%m-%d", errors='coerce')

identifiers.info()
```
