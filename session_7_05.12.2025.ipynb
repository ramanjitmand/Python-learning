```python
# import time module, and sleep module
import time

# Let's get numbers from 1 to 10
# Start : stop (1 after one we want): step
numbers = range(1,11,1)
numbers
# Result is range(1, 11) not an actual list

numbers = list(range(1,11,1))
numbers
```

```python
# Let's see loops in use
# The idea with a loop is if something has parts to it (e.g. lists, dictionaries) (not dataframes etc) it allows us to go through the parts or use parts to do something else
# Iterable - you can go over, a part at a time; e.g. lists are iterable, 

# Type 'for' to tell Python you're starting a loop
# 'in' says what's the little bit coming from
# Put : afterwards
# Underneath : start typing the loop (indentation is 1 tab or 4 spaces)
for number in numbers:
    print(number)
    # 1 means 1 second; print then sleep for 1 second
    time.sleep(1)
    # It goes through list bit by bit
```

```python
# iteratable is i here, or it could be called j or dog or whatever you want
# Iterator is i
for i in numbers:
    # Let's do the number squared
    outcome = i ** 2
    print(outcome)
    time.sleep(1)
    print("Inside the loop")
    # Stuff which is part of the loop has to be in indentation

print('Outside loop')
```

```python
# Let's put loops in list by making 2 lists
# Let's times together every odd and even number between 1 and 12

evens = list(range(2,13,2))
evens

odds = list(range(1,12,2))
odds

multiple_list = []

for even in evens:
    for odd in odds: # this is a nested loop (goes through internal loop before moving onto the external loop)
        result = even * odd
        stmt = f'{even} times {odd} = {result}'
        print(stmt)
        multiple_list.append(stmt)

print(multiple_list)
```
