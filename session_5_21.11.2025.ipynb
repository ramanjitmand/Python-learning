```python
# Session 5 starts here (at 13.32hrs)

# It's installed but pandas needs to be imported in
import pandas as pd

filepath = "https://github.com/data-to-insight/ERN-sessions/raw/main/data/903_xlsx_errors.xlsx"
# Simple way - dfs = pd.read_excel(filepath)
# Let's read in every table in excel by doing the following (None must have capital N at start)
dfs = pd.read_excel(filepath, sheet_name=None)
# Put in 'header' just for header tab, but can include multiple tabs too
# dfs
# This dataframe is 3 years old so will look different nowadays lol

# dfs.keys()

# You will see top 5 rows here
# Good idea to have Excel open on another screen :)
dfs['header'].head()
```

```python
# Rule 102 in 903 validator says date of birth is not a valid date

# Dates have to be formatted as d/m/y for 903 return
# to_datetime() (i.e. to_datetime function) - you can format argument e.g. "%d/%m/%Y" (2 digit day and month and 4 digit year)
# errors = "coerce" <-- This will return an NA for every row where date wasn't formatted correctly

header = dfs['header']

# Let's add a new column called DOB_dt (let's not overwrite original DOB column - let's do nondestructive work!)
header['DOB_dt'] = pd.to_datetime(header['DOB'], format="%d/%m/%Y", errors='coerce')

invalid_dobs = header['DOB_dt'].isna()
# True if there's an NA 

error_rows = header[invalid_dobs]
error_rows
```

```python
# Group exercise
# Rule 115 - Date of LA's decision that a child should be placed for adoption is not a valid date
# table = placed_for_adoption, col = DATE_PLACED

# I think we're calling the placed_for_adoption table p4a for short (in my attempt I was missing this bit)
p4a = dfs['placed_for_adoption']
# I think below means let's add a new DATE_PLACED_dt column which is based on original DATE_PLACED column giving date in dd/mm/yyyy format with errors given as NA
p4a['DATE_PLACED_dt'] = pd.to_datetime(p4a['DATE_PLACED'], format="%d/%m/%Y", errors='coerce')
# Now let's try to see the NAs in the new DATE_PLACED_dt column
# You can do it in 1 line as below or in 2 lines (like in example WLP did above?)
invalid_dp = p4a[p4a['DATE_PLACED_dt'].isna()]
invalid_dp
```

```python
# Let's look at something new now!

# Tip for me - If I close my codespace and reopen it I have to Run All! 

# Rule 103 - The ethnicity code is either not valid ot has not been entered
# header, ETHNIC - emmpty or not a valid code

# Valid ethnicity codes are as follows: 
# 103 - The ethnicity code is either not valid or has not been entered.
# header, ETHNIC - empty, or not a valid code

ethnicity_codes = [
        "WBRI",
        "WIRI",
        "WOTH",
        "WIRT",
        "WROM",
        "MWBC",
        "MWBA",
        "MWAS",
        "MOTH",
        "AIND",
        "APKN",
        "ABAN",
        "AOTH",
        "BCRB",
        "BAFR",
        "BOTH",
        "CHNE",
        "OOTH",
        "REFU",
        "NOBT",
    ]

# .isin()

error_rows = header[~header['ETHNIC'].isin(ethnicity_codes)]

error_rows
```

```python
# Group exercise 
# Rule 114 - Data entry to record the status of former carer(s) of an adopted child is invalid.
# table - ad1
# FOSTER_CARE
# valid codes 0, 1

ad1 = dfs['ad1']
# Everything in FOSTER_CARE column is a string
ad1['FOSTER_CARE'] = ad1['FOSTER_CARE'].astype('str')
# .isin looks at strings
error_rows = ad1[~ad1['FOSTER_CARE'].isin(["0","1"])]
error_rows
```
